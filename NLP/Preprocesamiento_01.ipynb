<font color='yellow'>Antes de ejecutar la siguiente celda</font>

En Python, anteponer la letra `"r"` a una cadena (por ejemplo, `r"\n"`) indica que se trata de una cadena en bruto o raw string. Esto significa que los caracteres de escape (como `\n`, `\t`, etc.) se interpretan literalmente, en lugar de procesarse como saltos de línea, tabulaciones, etc.

```python
"""
Script que descarga los recursos de NLTK necesarios y
aplica un patrón de expresión regular a una cadena de ejemplo.
"""

# Descarga de recursos de NLTK que se pueden requerir (en este caso wordnet y punkt)
nltk.download('wordnet')
nltk.download('punkt')

# Cadena de ejemplo sobre la que vamos a realizar la búsqueda
mi_cadena = "Hagamos 1 prueba de Detección de 12 expresiones regulares!"

# Definición del patrón para buscar secuencias de caracteres alfanuméricos
# \w  -> representa caracteres alfanuméricos (letras, dígitos y guion bajo)
# +   -> indica 'uno o más' de esos caracteres
patron = r"\w+"

# Ejemplos de otros patrones de interés que se podrían utilizar:
#   - r"[0-9]"        -> Coincide con cualquier dígito (0 al 9).
#   - r"\s+"          -> Coincide con uno o más espacios en blanco.
#   - r"[a-z]"        -> Coincide con letras minúsculas de la 'a' a la 'z'.
#   - r"[^a-z]"       -> Coincide con cualquier carácter que no sea letra minúscula.
#   - r"[A-Z]"        -> Coincide con letras mayúsculas de la 'A' a la 'Z'.
#   - r"[^0-9]"       -> Coincide con cualquier carácter que no sea dígito.
#   - r"\."           -> Coincide con el punto literal '.'
#   - r"\.$"          -> Coincide con un punto literal al final de la cadena.
#
# Para usar uno de los patrones anteriores, simplemente asigne su valor a 'patron'.
# Ejemplo: patron = r"[0-9]"

# Uso de findall para buscar todas las apariciones del patrón en la cadena
coincidencias = re.findall(patron, mi_cadena)

# Impresión de los resultados
print("Cadena original:")
print(mi_cadena)
print("\nPatrón utilizado:", patron)
print("Coincidencias encontradas:", coincidencias)
```

```python
# Veamos otros ejemplos de patrones

mi_cadena = "Hagamos 1 prueba * buenísima ¿? de Detección de 12 expresiones regulares en este código v2.0. Let's gooooo!"
print("Cadena original:")
print(mi_cadena)

# 1) Buscar dígitos (0-9)
patron = r"\d+"
coincidencias = re.findall(patron, mi_cadena)
print("\nPatrón utilizado:", patron)
print("Coincidencias encontradas:", coincidencias)

# 2) Buscar letras mayúsculas
patron = r"[A-Z]"
coincidencias = re.findall(patron, mi_cadena)
print("\nPatrón utilizado:", patron)
print("Coincidencias encontradas:", coincidencias)

# 3) Buscar secuencias de espacios en blanco
patron = r"\s+"
coincidencias = re.findall(patron, mi_cadena)
print("\nPatrón utilizado:", patron)
print("Coincidencias encontradas:", coincidencias)

# 4) Buscar palabras que comiencen con vocal
#    - \b indica un límite de palabra,
#    - [aeiouAEIOU] significa "cualquier vocal mayúscula o minúscula".
patron = r"\b[aeiouAEIOU]\w*"
coincidencias = re.findall(patron, mi_cadena)
print("\nPatrón utilizado:", patron)
print("Coincidencias encontradas:", coincidencias)

# 5) Buscar palabras de longitud fija; por ejemplo, 9 letras
patron = r"\b[a-zA-Z]{9}\b"
coincidencias = re.findall(patron, mi_cadena)
print("\nPatrón utilizado:", patron)
print("Coincidencias encontradas:", coincidencias)
print("Este patrón no reconoce palabras con tilde (p.ej.: /buenísima/), ya que el patrón no incluye caracteres con tilde\n")
print('Cambiamos el patrón por este otro r"\\b[a-zA-Záéíóú]{9}\\b" que sí incluye las palabras con tilde')
patron = r"\b[a-zA-Záéíóú]{9}\b"
coincidencias = re.findall(patron, mi_cadena)
print("\nPatrón utilizado:", patron)
print("Coincidencias encontradas:", coincidencias)

# 6) Buscar signos de puntuación o caracteres no alfanuméricos distintos de espacios
patron = r"[^\w\s]+"
coincidencias = re.findall(patron, mi_cadena)
print("\nPatrón utilizado:", patron)
print("Coincidencias encontradas:", coincidencias)

# 7) Buscar cualquier carácter seguido inmediatamente de un dígito
patron = r".\d"
coincidencias = re.findall(patron, mi_cadena)
print("\nPatrón utilizado:", patron)
print("Coincidencias encontradas:", coincidencias)

# 8) Buscar mayúscula al inicio de la cadena
patron = r"^[A-Z]"
coincidencias = re.findall(patron, mi_cadena)
print("\nPatrón utilizado:", patron)
print("Coincidencias encontradas:", coincidencias)

# 9) Buscar secuencias de tres vocales consecutivas
patron = r"[aeiouAEIOU]{3}"
coincidencias = re.findall(patron, mi_cadena)
print("\nPatrón utilizado:", patron)
print("Coincidencias encontradas:", coincidencias)

# 10) Buscar la exclamación final
patron = r"!$"
coincidencias = re.findall(patron, mi_cadena)
print("\nPatrón utilizado:", patron)
print("Coincidencias encontradas:", coincidencias)
```

```python
"""
Ejemplo de uso de diferentes funciones de la biblioteca 're' (expresiones regulares) en Python.

En este código, se demuestra cómo:
- Definir una cadena de texto y un patrón (regex).
- Aplicar re.split(), re.match(), re.search() y re.findall() a esa cadena.
- Observar los distintos resultados que se obtienen de cada método.
"""

# Cadena de ejemplo sobre la que aplicaremos las expresiones regulares
mi_cadena = (
    "Hagamos 1 prueba * buenísima ¿? de Detección de 12 expresiones regulares "
    "en este código v2.0. Let's gooooo!"
)

# 1. Definir un patrón para split
#    Este patrón, por ejemplo, puede dividir la cadena cada vez que encuentra
#    un símbolo de puntuación o caracter especial (no alfanumérico).
#    Aquí se muestra un patrón arbitrario de ejemplo; puedes cambiarlo según tus necesidades.
PATRON_split = r"[^a-zA-Z0-9]+"

# Usamos re.split() para dividir la cadena según el patrón.
# Devuelve una lista con los fragmentos de cadena que coinciden con la "separación".
resultado_split = re.split(PATRON_split, mi_cadena)

# 2. Usar re.match() para comprobar si la cadena empieza con 'Hag'
#    - re.match(patrón, cadena) solo busca al inicio de la cadena.
#    - Si coincide, devuelve un objeto "Match", de lo contrario None.
resultado_match = re.match('Hag', mi_cadena)

# 3. Usar re.search() para buscar la primera aparición de 'pru' en la cadena
#    - re.search(patrón, cadena) busca en todo el texto hasta hallar la primera coincidencia.
#    - Devuelve un objeto "Match" o None si no encuentra coincidencia.
resultado_search = re.search('pru', mi_cadena)

# 4. Definir un patrón para encontrar dígitos o secuencias de caracteres alfanuméricos
#    - (\d+|\w+)
#      - \d+: uno o más dígitos
#      - \w+: una o más letras, dígitos o subrayados (equivale a [a-zA-Z0-9_]+)
#    - Los paréntesis capturan esos grupos.
PATRON_digitos_palabras = r'(\d+|\w+)'

# Usar re.findall() para devolver todas las coincidencias en la cadena.
# Este método devuelve una lista de todas las cadenas que coinciden con el patrón.
resultado_findall = re.findall(PATRON_digitos_palabras, mi_cadena)

# 5. Usar re.match() con un patrón distinto:
#    [0-9a-z ]+  -> uno o más caracteres que sean dígitos (0-9), letras minúsculas o espacio en blanco
#    - Observa que podría no coincidir al inicio, si la cadena no empieza con esos caracteres.
#    - En este caso, "H" es mayúscula, por lo que la coincidencia será None.
resultado_match_otro = re.match('[0-9a-z ]+', mi_cadena)

# Impresión de resultados
print("Cadena de ejemplo:")
print(mi_cadena, "\n")

print(f"1) re.split('{PATRON_split}', mi_cadena):")
print(resultado_split, "\n")

print("2) re.match('Hag', mi_cadena):")
print("   Sí hay coincidencia" if resultado_match else "   No hay coincidencia", "\n")

print("3) re.search('pru', mi_cadena):")
if resultado_search:
    print(f"   Coincidencia encontrada en el índice {resultado_search.start()} "
          f"hasta {resultado_search.end()}")
else:
    print("   Sin coincidencia")
print()

print(f"4) re.findall('{PATRON_digitos_palabras}', mi_cadena):")
print(resultado_findall, "\n")

print("5) re.match('[0-9a-z ]+', mi_cadena):")
print("   Coincidencia" if resultado_match_otro else "   Sin coincidencia")
```

## 🔹 Sección 2: Tokenización de texto a nivel de palabra

🔍 **Objetivo:**  
Aprender diferentes estrategias de **tokenización** en Python, un paso fundamental en el **preprocesamiento de texto para PLN**.

📌 **Introducción**

La **tokenización** es el proceso de dividir un texto en unidades más pequeñas llamadas *tokens*, que pueden ser palabras, frases o caracteres. Este es un paso fundamental en el **preprocesamiento de lenguaje natural (PLN)**, en tareas como **análisis de texto, clasificación, búsqueda de información y generación de lenguaje natural**, ya que permite estructurar los datos textuales para análisis posteriores.

| `word_tokenize()` | NLTK | Tokenización estándar de palabras y puntuación. |
| `TreebankWordTokenizer()` | NLTK | Usa reglas del Penn Treebank, mejor con contracciones. |

| `RegexpTokenizer()` | NLTK | Tokenización personalizada con expresiones regulares. |
| `split()` | Python | Divide palabras usando solo espacios. |

```python
"""
📌 Ejemplo de diferentes estrategias de tokenización en NLTK y otras bibliotecas de Python.
Cada método tiene su propia lógica y es útil según el contexto de procesamiento de texto.
"""

# Cargar el modelo de spaCy para procesamiento NLP
nlp_en = spacy.load("en_core_web_sm") # Esto carga un modelo en inglés
nlp_es = spacy.load("es_core_news_sm") # Esto carga un modelo en español
"""En la url (https://spacy.io/models/) se pueden consultar más modelos disponibles """

# 📌 Texto de prueba para los métodos de tokenización
text1 = "Hola a todos. Soy Julian. Esta es nuestra primera tokenización del curso 2025 - I. ¡Vamos a ver qué tan bien sale!"
text2 = "Hi you all. I'm Julian. Let's see how to tokenize words in this course 2025 - I. I hope you're doing well!"

# 📌 Ejemplo 1: Tokenización básica con `word_tokenize()`
tokens1 = word_tokenize(text1)
print("🔹 Tokenización con `word_tokenize()`:")
print(tokens1)
print("\n" + "-"*50 + "\n")

# 📌 Ejemplo 2: Tokenización con `TreebankWordTokenizer`
tokenizer_treebank = TreebankWordTokenizer()
tokens2 = tokenizer_treebank.tokenize(text1)
print("🔹 Tokenización con `TreebankWordTokenizer()`:")
print(tokens2)
print("\n" + "-"*50 + "\n")

# 📌 Ejemplo 3: Tokenización con `WordPunctTokenizer`
tokenizer_word_punct = WordPunctTokenizer()
tokens3 = tokenizer_word_punct.tokenize(text2)
print("🔹 Tokenización con `WordPunctTokenizer()`:")
print(tokens3)
print("\n" + "-"*50 + "\n")

# 📌 Ejemplo 4: Tokenización con `RegexpTokenizer`
tokenizer_regex = RegexpTokenizer(r"[\w']+")
tokens4 = tokenizer_regex.tokenize(text2)
print("🔹 Tokenización con `RegexpTokenizer()`:")
print(tokens4)
print("\n" + "-"*50 + "\n")

# 📌 Ejemplo 5: Tokenización con `split()` de Python (Básico)
tokens5 = text2.split()
print("🔹 Tokenización con `split()`:")
print(tokens5)
print("\n" + "-"*50 + "\n")

# 📌 Ejemplo 6: Tokenización con `spaCy` en español
doc = nlp_es(text1)
tokens6 = [token.text for token in doc]
print("🔹 Tokenización con `spaCy` en español:")
print(tokens6)
print("\n" + "-"*50 + "\n")

# 📌 Ejemplo 7: Tokenización con `spaCy` en inglés
doc = nlp_en(text2)
tokens7 = [token.text for token in doc]
print("🔹 Tokenización con `spaCy` en inglés:")
print(tokens7)
print("\n" + "-"*50 + "\n")

# 📌 Ejemplo 8a: Tokenización con `TextBlob` en español
blob = TextBlob(text1)
tokens8a = blob.words
print("🔹 Tokenización con `TextBlob` en español:")
print(tokens8a)
print("\n" + "-"*50 + "\n")

# 📌 Ejemplo 8b: Tokenización con `TextBlob` en inglés
blob = TextBlob(text2)
tokens8b = blob.words
print("🔹 Tokenización con `TextBlob` en inglés:")
print(tokens8b)
print("\n" + "-"*50 + "\n")

# 📌 Ejemplo 9: MWETokenizer - Manejo de expresiones multi-palabra
text3 = "En esta clase de Procesamiento de Lenguaje Natural veremos algunas técnicas de inteligencia artificial aplicadas."
tokenizer_mwe = MWETokenizer([('Lenguaje', 'Natural'), ('inteligencia', 'artificial')])
tokens9 = tokenizer_mwe.tokenize(text3.split())
print("🔹 Tokenización con `MWETokenizer`:")
print(tokens9)
print("\n" + "-"*50 + "\n")

# 📌 Ejemplo 10: BlanklineTokenizer - usa líneas en blanco como separador.
# Esta técnica de tokenización puede ser útil cuando queremos tokenizar a nivel de frase.
text_multiline = "Primera línea.\n\nSegunda línea después de un espacio.\n\nTercera línea."
tokenizer_blankline = BlanklineTokenizer()
tokens10 = tokenizer_blankline.tokenize(text_multiline)
print("🔹 Tokenización con `BlanklineTokenizer`:")
print(tokens10)
print("\n" + "-"*50 + "\n")

# 📌 Ejemplo 11: AutoTokenizar con Transformers de HuggingFace
tokenizer_en = AutoTokenizer.from_pretrained("bert-base-uncased") # Este es un buen modelo para inglés
tokenizer_es = AutoTokenizer.from_pretrained('MMG/xlm-roberta-base-sa-spanish') # Este es un buen modelo para español
#MMG/xlm-roberta-base-sa-spanish
#vgaraujov/bart-base-spanish
text_es = "¡Tokenización en este curso usando transformers no se ve tan prometedor! Vamos a hacerlo."
text_en = "Tokenization with transformers in this course is powerful! Let's do it."
tokens11a = tokenizer_es.tokenize(text_es)
print("🔹 Tokenización con `Hugging Face Transformers` en español:")
print(tokens11a)
tokens11b = tokenizer_en.tokenize(text_en)
print("🔹 Tokenización con `Hugging Face Transformers` en inglés:")
print(tokens11b)
print("\n" + "-"*50 + "\n")

# 📌 Ejemplo 12: Tokenizar con Stanza
nlp_stanza = stanza.Pipeline(lang="es", processors="tokenize")
text = "¡Hola mundo! Probemos la tokenización con Stanza."
# Procesar el texto
doc = nlp_stanza(text1)
# Extraer tokens
tokens12 = [word.text for sentence in doc.sentences for word in sentence.words]
print("🔹 Tokenización con `Stanza`:")
print(tokens12)
print("\n" + "-"*50 + "\n")

# 📌 Ejemplo 13: Tokenizar con Gensim
tokens13 = simple_preprocess(text1)
print("🔹 Tokenización con `Gensim`:")
print(tokens13)
print("\n" + "-"*50 + "\n")
```

### 🔎 Explicación de cada método de tokenización

### ✅ `word_tokenize()`
- Usa el **modelo Punkt de tokenización**.
- Separa palabras y signos de puntuación en tokens individuales.
- Maneja contracciones y apóstrofes con reglas predefinidas.

### ✅ `TreebankWordTokenizer()`
- Basado en el **Penn Treebank Tokenizer**.
- Detecta mejor las contracciones en inglés, como `"Let's"` → `["Let", "'s"]`.

### ✅ `WordPunctTokenizer()`
- Divide las palabras y la puntuación completamente.
- `"Let's"` se convierte en `["Let", "'", "s"]`.

### ✅ `RegexpTokenizer()`
- Permite definir patrones personalizados con **expresiones regulares**.
- `[\w']+` mantiene palabras y contracciones juntas.

### ✅ `split()` (Método básico de Python)
- Separa el texto **usando solo espacios en blanco**.
- No maneja puntuaciones o contracciones de manera eficiente.

### ✅ `spaCy Tokenizer`
- Procesa el texto con **modelos NLP avanzados**.
- Detecta nombres propios, entidades y estructuras complejas.

### ✅ `TextBlob Tokenizer`
- Tokeniza palabras de manera simple.
- Integra herramientas NLP adicionales.

### ✅ `Hugging Face Tokenizers`
- Tokenización optimizada para **modelos de deep learning**.
- Usa **subword tokenization**, lo que permite manejar palabras desconocidas y morfologías complejas.
- Compatible con **BERT, GPT, RoBERTa, T5** y otros modelos de PLN.

### ✅ `Stanza Tokenizer` (StanfordNLP)
- Tokenización basada en **redes neuronales** con análisis morfosintáctico.
- Compatible con múltiples idiomas y adaptado a textos con estructura gramatical compleja.
- Detecta tokens con una **comprensión contextual avanzada**.

### ✅ `Gensim Tokenizer`
- Optimizado para **procesamiento de texto en modelado de temas y embeddings**.
- `simple_preprocess()` elimina puntuación y convierte el texto a **tokens en minúsculas**.
- Ideal para modelos como **LDA y Word2Vec**.

## 🔹 Sección 3: Tokenización de Frases en PLN 📖  
### 🔍 ¿Qué es la Tokenización de Frases?

La **tokenización de frases** es el proceso de dividir un texto en **oraciones individuales** en lugar de palabras.  
Es un paso clave en el preprocesamiento de texto para tareas como:

✅ **Análisis de Sentimiento**: Evaluar la polaridad de cada oración.  
✅ **Resumen Automático**: Extraer oraciones clave en textos largos.  
✅ **Traducción Automática**: Segmentar frases antes de la traducción.  
✅ **Extracción de Información**: Dividir documentos en unidades más manejables.

En este cuaderno exploraremos **diferentes enfoques** de tokenización de frases usando varias bibliotecas de **Python para PLN**.

```python
"""
📌 Ejemplo de diferentes estrategias de tokenización de frases en Python.
Cada método tiene su propia lógica y es útil según el contexto de procesamiento de texto.
"""

# 📌 Texto de prueba
text = """Hola a todos. ¡Bienvenidos al curso de PLN!
La tokenización de frases es importante. ¡Esto es solo un ejemplo!"""

# 1️⃣ Tokenización con `sent_tokenize()` de NLTK
tokens_nltk = sent_tokenize(text)
print("🔹 Tokenización con `sent_tokenize()` de NLTK:")
print(tokens_nltk)
print("\n" + "-"*50 + "\n")

# 2️⃣ Tokenización con `split('.')` de Python (básico)
tokens_split = text.split('.')
tokens_split = [sentence.strip() for sentence in tokens_split if sentence.strip()]  # Eliminar espacios vacíos
print("🔹 Tokenización con `split('.')`:")
print(tokens_split)
print("\n" + "-"*50 + "\n")

# 3️⃣ Tokenización con `spaCy`
doc = nlp_es(text)
tokens_spacy = [sent.text for sent in doc.sents]
print("🔹 Tokenización con `spaCy`:")
print(tokens_spacy)
print("\n" + "-"*50 + "\n")

# 4️⃣ Tokenización con `TextBlob`
blob = TextBlob(text)
tokens_textblob = blob.sentences
print("🔹 Tokenización con `TextBlob`:")
print([str(sent) for sent in tokens_textblob])
print("\n" + "-"*50 + "\n")

# 5️⃣ Tokenización con `Stanza`
doc_stanza = nlp_stanza(text)
tokens_stanza = [sentence.text for sentence in doc_stanza.sentences]
print("🔹 Tokenización con `Stanza`:")
print(tokens_stanza)
```

### 🔎 **Explicación de cada método de tokenización de frases**

### ✅ `sent_tokenize()` (NLTK)
- Usa modelos preentrenados basados en **Punkt** para segmentar frases.
- Funciona bien en textos estructurados y es rápido.
- No siempre maneja abreviaciones correctamente.

### ✅ `split('.')` (Python)
- Divide el texto usando el **punto (.)** como delimitador.
- Es **rápido** pero **no detecta abreviaciones** (`Dr.`, `etc.`) y deja frases cortadas.

### ✅ `spaCy Sentence Tokenizer`
- Tokenización basada en **modelos NLP avanzados**.
- **Detecta abreviaciones y estructuras gramaticales complejas**.
- Compatible con múltiples idiomas.

### ✅ `TextBlob Sentence Tokenizer`
- Basado en NLTK pero con integración a herramientas NLP adicionales.
- Útil cuando se necesita tokenización con análisis de sentimientos.

### ✅ `Stanza Sentence Tokenizer`
- Basado en **redes neuronales** y modelos gramaticales.
- **Maneja diferentes idiomas y estructuras de texto complejas**.

| `split('.')` | Baja | No detecta abreviaciones | Casos básicos |

## 🔹 Sección 4: Lematización y Stemming en PLN 📖  
### 🔍 ¿Qué son la Lematización y el Stemming?

La **lematización** y el **stemming** son técnicas de normalización de texto en **Procesamiento de Lenguaje Natural (PLN)**.  
Ambas reducen las palabras a su forma base, pero con enfoques distintos:

- 🌱 **Stemming**: Recorta palabras a su raíz sin considerar contexto ni reglas gramaticales.  
  - *Ejemplo:* `"correr", "corriendo", "corrió"` → `"corr"`  
- 🍃 **Lematización**: Transforma palabras a su **forma canónica** considerando su significado y contexto gramatical.  
  - *Ejemplo:* `"correr", "corriendo", "corrió"` → `"correr"`  

💡 Estas técnicas son útiles en tareas como:
✅ **Análisis de sentimiento**  
✅ **Búsqueda de información (IR)**  
✅ **Reducción de dimensionalidad en PLN**  
✅ **Clasificación de texto**  

En este cuaderno exploraremos **diferentes enfoques de lematización y stemming en Python**, aplicándolos **a nivel de párrafos**.

---

### 📌 **Métodos que exploraremos en este cuaderno**  

| 🛠️ **Método** | 🔍 **Librería** | 📌 **Descripción** |
|--------------|--------------|------------------|
| `PorterStemmer` | NLTK | Stemming basado en reglas para inglés. |
| `SnowballStemmer` | NLTK | Stemming multilingüe con mejoras sobre Porter. |
| `WordNetLemmatizer` | NLTK | Lematización basada en el diccionario de WordNet. |
| `spaCy Lemmatizer` | spaCy | Lematización avanzada con modelos NLP. |
| `Stanza Lemmatizer` | Stanza | Lematización con redes neuronales y análisis morfosintáctico. |

---



```python
"""
📌 Ejemplo de lematización y stemming en un párrafo usando diferentes librerías en Python.
Cada método tiene su propia lógica y es útil según el contexto de procesamiento de texto.
"""

# 📌 Párrafo de prueba
text = """Al amanecer, un grupo de estudiantes subía lentamente la colina para
meditar antes de sus exámenes parciales.
Respiraban profundo, tratando de calmar sus mentes antes de enfrentarse a las
complejas preguntas que los esperaban.
La brisa matutina y el canto de los pájaros acompañaban su breve momento de
concentración, ayudándolos a despejar sus pensamientos y fortalecer su confianza."""

# 📌 Tokenización de frases y palabras
sentences = sent_tokenize(text)
words = [word_tokenize(sent) for sent in sentences]

# 1️⃣ **Stemming con PorterStemmer (NLTK)**
stemmer_porter = PorterStemmer()
stemmed_porter = [[stemmer_porter.stem(word) for word in sentence] for sentence in words]
print("🔹 Stemming con `PorterStemmer()` de NLTK:")
print(stemmed_porter)
print("\n" + "-"*50 + "\n")

# 2️⃣ **Stemming con SnowballStemmer (NLTK)**
stemmer_snowball = SnowballStemmer("spanish")
stemmed_snowball = [[stemmer_snowball.stem(word) for word in sentence] for sentence in words]
print("🔹 Stemming con `SnowballStemmer()` de NLTK:")
print(stemmed_snowball)
print("\n" + "-"*50 + "\n")

# 3️⃣ **Lematización con WordNetLemmatizer (NLTK)**
lemmatizer_nltk = WordNetLemmatizer()
lemmatized_nltk = [[lemmatizer_nltk.lemmatize(word) for word in sentence] for sentence in words]
print("🔹 Lematización con `WordNetLemmatizer()` de NLTK:")
print(lemmatized_nltk)
print("\n" + "-"*50 + "\n")

# 4️⃣ **Lematización con spaCy**
doc_spacy = nlp_es(text)
lemmatized_spacy = [[token.lemma_ for token in sent] for sent in doc_spacy.sents]
print("🔹 Lematización con `spaCy`:")
print(lemmatized_spacy)
print("\n" + "-"*50 + "\n")

# 5️⃣ **Lematización con Stanza**
nlp_stanza = stanza.Pipeline(lang="es", processors="tokenize,lemma")
doc_stanza = nlp_stanza(text)
lemmatized_stanza = [[word.lemma for word in sentence.words] for sentence in doc_stanza.sentences]
print("🔹 Lematización con `Stanza`:")
print(lemmatized_stanza)
```

### 🔎 **Explicación de cada método de lematización y stemming**

### ✅ `PorterStemmer()` (NLTK)
- Aplica **Stemming basado en reglas**.
- Funciona bien en inglés, pero puede generar palabras truncadas.
- No siempre devuelve palabras significativas.

### ✅ `SnowballStemmer()` (NLTK)
- Variante mejorada de Porter con soporte multilingüe.
- Maneja mejor **palabras en español**.
- Aún puede truncar palabras sin significado.

### ✅ `WordNetLemmatizer()` (NLTK)
- Usa el **diccionario WordNet** para lematización precisa.
- Requiere especificar la categoría gramatical (`pos`).
- Más preciso que el stemming pero limitado a inglés.

### ✅ `spaCy Lemmatizer`
- **Lematización avanzada** basada en modelos NLP.
- Maneja diferentes idiomas y categorías gramaticales.
- Precisión alta, pero requiere cargar un modelo específico.

### ✅ `Stanza Lemmatizer`
- **Lematización con redes neuronales** y análisis morfosintáctico.
- **Maneja múltiples idiomas** con alta precisión.
- Requiere descargar modelos para cada idioma.

---

## 📌 **Resumen de los métodos de lematización y stemming**

| Método | Basado en | Idiomas | Precisión | Casos de uso |
|--------|----------|---------|-----------|--------------|
| `PorterStemmer()` | Reglas | Inglés | Baja | Rápido pero impreciso |
| `SnowballStemmer()` | Reglas | Español/Inglés | Media | Mejor para textos en español |
| `WordNetLemmatizer()` | WordNet | Inglés | Alta | Necesita especificar POS |
| `spaCy` | Modelos NLP | Español/Inglés | Muy alta | Procesamiento NLP avanzado |
| `Stanza` | Redes neuronales | Español/Inglés | Muy alta | NLP con análisis sintáctico |

---

## 🚀 **Conclusión**
La elección entre **stemming y lematización** depende del caso de uso.  
Si necesitas rapidez, **SnowballStemmer** es una opción básica.  
Si buscas **precisión lingüística**, usa **spaCy o Stanza**.  

## 🔹 Sección 5: Extracción de Inventarios de Términos en PLN  
### 🔍 ¿Qué es la Extracción de Diccionarios?

La **extracción de diccionarios** en Procesamiento de Lenguaje Natural (PLN) consiste en identificar y recopilar términos clave dentro de un texto.  
Estos inventarios de términos pueden utilizarse en:

✅ **Clasificación de documentos** (extracción de palabras clave para categorización).  
✅ **Generación de glosarios** (creación de diccionarios temáticos).  
✅ **Detección de entidades** (nombres propios, lugares, fechas relevantes).  
✅ **Análisis de contenido** (palabras más frecuentes en un corpus de texto).  

En este cuaderno exploraremos **diferentes enfoques para extraer términos a nivel de párrafos**, aplicando técnicas de tokenización, filtrado y conteo.

---

### 📌 **Métodos que exploraremos en este cuaderno**  

| 🛠️ **Método** | 🔍 **Librería** | 📌 **Descripción** |
|--------------|--------------|------------------|
| `Counter()` | Python (`collections`) | Crea un inventario de términos basado en frecuencia. |
| `TfidfVectorizer` | Scikit-learn | Identifica términos clave mediante peso TF-IDF. |
| `spaCy NER` | spaCy | Extrae entidades nombradas (nombres, fechas, lugares). |
| `YAKE!` | PKE (Pipeline de PLN) | Método no supervisado para extracción de palabras clave. |

---


```python
"""
📌 Ejemplo de extracción de diccionarios de términos a nivel de párrafos.
Cada método captura términos clave mediante diferentes enfoques.
"""

# 📌 Párrafo de prueba
text = """Al amanecer, un grupo de estudiantes subía lentamente la colina para meditar antes de sus exámenes parciales.
Respiraban profundo, tratando de calmar sus mentes antes de enfrentarse a las complejas preguntas que los esperaban.
La brisa matutina y el canto de los pájaros acompañaban su breve momento de concentración, ayudándolos a despejar sus pensamientos y fortalecer su confianza."""

# 1️⃣ **Método con `Counter()` (frecuencia de palabras)**
tokens = word_tokenize(text.lower())  # Convertir a minúsculas y tokenizar
filtered_tokens = [word for word in tokens if word.isalnum()]  # Filtrar solo palabras alfanuméricas
word_freq = Counter(filtered_tokens)

print("🔹 Inventario de términos con `Counter()`:")
print(word_freq.most_common(10))  # Mostrar las 10 palabras más frecuentes
print("\n" + "-"*50 + "\n")

# 2️⃣ **Método con `TfidfVectorizer()` (importancia de términos)**

spanish_stop_words = ["de", "la", "el", "en", "y", "a", "los", "las", "un", "una", "con", "para", "por", "es", "está", "son", "fue", "como", "más", "pero"]

vectorizer = TfidfVectorizer(stop_words=spanish_stop_words)
tfidf_matrix = vectorizer.fit_transform([text])  # Ajustar vectorizador al texto
tfidf_scores = dict(zip(vectorizer.get_feature_names_out(), tfidf_matrix.toarray()[0]))

# Ordenar términos por importancia TF-IDF
sorted_tfidf = sorted(tfidf_scores.items(), key=lambda x: x[1], reverse=True)[:10]

print("🔹 Inventario de términos con `TfidfVectorizer()`:")
print(sorted_tfidf)
print("\n" + "-"*50 + "\n")

# 3️⃣ **Método con `spaCy` (entidades nombradas)**
doc_spacy = nlp_es(text)
named_entities = [(ent.text, ent.label_) for ent in doc_spacy.ents]

print("🔹 Inventario de términos con `spaCy` (NER - entidades nombradas):")
print(named_entities)
print("\n" + "-"*50 + "\n")

# 4️⃣ **Método con `YAKE!` (palabras clave)**
kw_extractor = yake.KeywordExtractor(lan="es", n=2, top=10)
keywords = kw_extractor.extract_keywords(text)

print("🔹 Inventario de términos con `YAKE!`:")
print(keywords)
```

### 🔎 **Explicación de cada método de extracción de términos**

### ✅ `Counter()` (Frecuencia de Palabras)
- Cuenta cuántas veces aparece cada palabra en el texto.
- **Ventajas**: Rápido y sencillo.
- **Limitaciones**: No distingue términos importantes, solo mide frecuencia.

### ✅ `TfidfVectorizer()` (TF-IDF)
- Evalúa la importancia de una palabra en función de su frecuencia en el documento y en el corpus.
- **Ventajas**: Destaca términos clave en función de su relevancia.
- **Limitaciones**: No identifica palabras con significado contextual.

### ✅ `spaCy NER` (Entidades Nombradas)
- Extrae nombres propios, fechas, lugares y organizaciones.
- **Ventajas**: Detecta entidades clave en el texto.
- **Limitaciones**: No extrae términos genéricos importantes.

### ✅ `YAKE!` (Extracción de Palabras Clave)
- Algoritmo no supervisado que encuentra palabras clave en textos sin necesidad de entrenamiento.
- **Ventajas**: Funciona sin necesidad de un corpus de referencia.
- **Limitaciones**: Puede no capturar algunos términos relevantes si el texto es muy corto.

---

### 📌 **Resumen de los métodos de extracción de términos**

| Método | Basado en | Precisión | Casos de uso |
|--------|----------|-----------|--------------|
| `Counter()` | Frecuencia | Baja | Análisis básico de palabras |
| `TfidfVectorizer()` | Frecuencia y relevancia | Alta | Extracción de términos clave |
| `spaCy NER` | Modelos NLP | Muy alta | Extracción de entidades nombradas |
| `YAKE!` | Algoritmos no supervisados | Alta | Extracción automática de palabras clave |

---

### 🚀 **Conclusión**
Cada método de extracción de diccionarios tiene ventajas y desventajas según el tipo de análisis.  
Si buscas rapidez, **Counter()** es útil.  
Si necesitas términos más relevantes, **TF-IDF o YAKE!** funcionan mejor.  
Para extraer nombres propios y entidades, **spaCy NER** es la mejor opción.
