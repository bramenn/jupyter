## 🔹 Sección 1: Stopwords (Definición e Importancia)

#### 🔍 **Objetivo:**
Comprender el rol de las **stopwords** (*palabras vacías*) en **Procesamiento de Lenguaje Natural (PLN)**, cómo su eliminación puede **mejorar** la calidad de nuestros datos y aprender distintas formas de implementarlo en Python usando **NLTK**, **spaCy**, **TextBlob** y **scikit-learn**.

---

#### 📌 **Introducción**

Las **stopwords** son términos que, en la mayoría de los casos, aportan **poca información semántica** (por ejemplo, artículos, conjunciones o preposiciones).  
- En español, podemos mencionar: *el, la, de, y, a*, entre otras.  
- En inglés, son comunes: *the, of, and, in, to*, etc.

Al carecer de información relevante sobre el **contenido** del texto, estas palabras suelen eliminarse para:
- ✅ **Reducir el ruido** y facilitar el modelado de texto.  
- ✅ **Mejorar el desempeño** en tareas de **clasificación** y **análisis de sentimientos**.  
- ✅ **Acelerar** el procesamiento al disminuir la cantidad de tokens.

No obstante, **no siempre** conviene descartarlas. Dependiendo del objetivo (por ejemplo, *análisis sintáctico* o *construcción de oraciones*), se podría perder **información** valiosa.

---

#### 📌 **Métodos que exploraremos en este cuaderno**

A continuación se describen algunas aproximaciones para gestionar **stopwords** en Python, haciendo énfasis en:
1. **NLTK**: Manejo de stopwords en distintos idiomas, utilizando las listas de `nltk.corpus.stopwords`.
2. **spaCy**: Uso de `token.is_stop` y filtros de spaCy según el idioma.
3. **TextBlob**: Creación de listas personalizadas para filtrar tokens.
4. **scikit-learn (TfidfVectorizer)**: Integración directa del filtrado de stopwords en la fase de vectorización.

| 🔢 **Método**           | 🛠️ **Función / Módulo**                          | 📌 **Descripción**                                                                   |
|-------------------------|--------------------------------------------------|---------------------------------------------------------------------------------------|
| **Listas de NLTK**      | `nltk.corpus.stopwords`                          | Contiene listas de stopwords en distintos idiomas.                                   |
| **spaCy**               | `token.is_stop`                                  | Detección y filtrado de palabras vacías basadas en el modelo lingüístico.            |
| **TextBlob**            | Uso de listas personalizadas                     | TextBlob no trae una lista extensa por defecto, se crea manualmente.                |
| **scikit-learn**        | `TfidfVectorizer(stop_words='english' o lista)` | Permite eliminar stopwords durante la vectorización, optimizando modelos de ML.      |
| **Filtrado de tokens**  | `if token not in stopwords`                      | Proceso básico para excluir palabras irrelevantes.                                   |
| **Tokenización (NLTK)** | `nltk.tokenize.word_tokenize()`                  | Divide el texto en palabras individuales para su procesamiento posterior.           |

A continuación, se presentan ejemplos que ilustran cómo filtrar stopwords con **NLTK**, **spaCy**, **TextBlob** y **scikit-learn**. El código que sigue muestra, paso a paso, la utilización de cada enfoque en **español** e **inglés**:

```python
""" ------------------------------------------------------------------------ """
"""  📌 Usando NLTK """
""" ------------------------------------------------------------------------ """

# ---
# 1️⃣ Ejemplo en español (NLTK)
# ---

texto_es = """
En el campo del Procesamiento de Lenguaje Natural, la eliminación de stopwords
puede mejorar el rendimiento en diversas tareas.
"""

tokens_es = word_tokenize(texto_es, language='spanish')
stopwords_es = set(stopwords.words('spanish'))

print("Las stop-words en español que tiene NLTK son: ", stopwords_es)
print(f"\nEl total de stop-words en ESPAÑOL es: {len(stopwords_es)}\n")
tokens_es_filtrados = [
    t for t in tokens_es
    if t.lower() not in stopwords_es and t.isalpha()
]

print("\n--- EJEMPLO STOPWORDS EN ESPAÑOL ---")
print("Tokens originales (ES):", tokens_es)
print("Tokens sin stopwords (ES):", tokens_es_filtrados,'\n')


# ---
# 2️⃣ Ejemplo en inglés (NLTK)
# ---

texto_en = """
In the field of Natural Language Processing, removing stopwords
can improve performance in many tasks.
"""

tokens_en = word_tokenize(texto_en, language='english')
stopwords_en = set(stopwords.words('english'))

print("Las stop-words en inglés que tiene NLTK son: ", stopwords_en)
print(f"\nEl total de stop-words en INGLÉS es: {len(stopwords_es)}\n")

tokens_en_filtrados = [
    t for t in tokens_en
    if t.lower() not in stopwords_en and t.isalpha()
]

print("\n--- EJEMPLO STOPWORDS EN INGLÉS ---")
print("Tokens originales (EN):", tokens_en)
print("Tokens sin stopwords (EN):", tokens_en_filtrados)
```

```python
""" ------------------------------------------------------------------------ """
""" 📌 Usando Spacy """
""" ------------------------------------------------------------------------ """

texto_es_spacy = """
En el campo del Procesamiento de Lenguaje Natural, la eliminación de stopwords
puede mejorar el rendimiento en diversas tareas.
"""
texto_en_spacy = """
In the field of Natural Language Processing, removing stopwords
can improve performance in many tasks.
"""

# Procesamiento con spaCy
doc_es = nlp_es(texto_es_spacy)
doc_en = nlp_en(texto_en_spacy)

# Cargamos el listado de stop-words que trae Spacy tanto en español como en
# inglés
stop_words_es = nlp_es.Defaults.stop_words
stop_words_en = nlp_en.Defaults.stop_words

# ---------------------------------------------
# Las mostramos para hacernos una idea
print("Stop words en español en spaCy:")
for word in sorted(stop_words_es):
    print(word)

# Mostramos el conteo total
print(f"\nEl total de stop-words en ESPAÑOL es: {len(stop_words_es)}\n")
# ---------------------------------------------
# Print all stop words
print("Stop words en inglés en spaCy:")
for word in sorted(stop_words_en):
    print(word)

# Mostramos el conteo total
print(f"\nEl total de stop-words en INGLÉS es: {len(stop_words_en)}\n")
# ---------------------------------------------

# Luego hacemos el filtrado de tokens
tokens_es_spacy_filtrados = [token.text for token in doc_es if not token.is_stop and token.is_alpha]
tokens_en_spacy_filtrados = [token.text for token in doc_en if not token.is_stop and token.is_alpha]

# El atributo token.is_stop permite filtrar fácilmente las stopwords definidas por spaCy.
# Filtramos además tokens que sean alfabéticos, evitando puntuación y caracteres extraños (token.is_alpha).

print("--- EJEMPLO STOPWORDS EN ESPAÑOL (spaCy) ---")
print("Tokens sin stopwords (ES):", tokens_es_spacy_filtrados)

print("\n--- EJEMPLO STOPWORDS EN INGLÉS (spaCy) ---")
print("Tokens sin stopwords (EN):", tokens_en_spacy_filtrados)
```

```python
""" ------------------------------------------------------------------------ """
""" 📌 Usando TextBlob """
""" ------------------------------------------------------------------------ """

# No trae por defecto muchas opciones de stopwords, pero podemos crear nuestra propia lista

# custom_stopwords = {"en", "to"}
custom_stopwords = {"en", "el", "la", "de", "y", "a", "the", "in", "and", "of", "to"}

texto_tb_es = """
En el campo del Procesamiento de Lenguaje Natural, la eliminación de stopwords
puede mejorar el rendimiento en diversas tareas.
"""

blob_es = TextBlob(texto_tb_es)
words_es = blob_es.words  # Lista de palabras
filtered_words_es = [word for word in words_es if word.lower() not in custom_stopwords]

print("\n--- EJEMPLO STOPWORDS EN ESPAÑOL (TextBlob) ---")
print("Tokens originales (ES):", words_es)
print("Tokens sin stopwords (ES):", filtered_words_es)


texto_tb_en = """
In the field of Natural Language Processing, removing stopwords
can improve performance in many tasks.
"""

blob_en = TextBlob(texto_tb_en)
words_en = blob_en.words  # Lista de palabras
filtered_words_en = [word for word in words_en if word.lower() not in custom_stopwords]

print("\n\n--- EJEMPLO STOPWORDS EN INGLÉS (TextBlob) ---")
print("Tokens originales (EN):", words_en)
print("Tokens sin stopwords (EN):", filtered_words_en)
```

```python
""" ------------------------------------------------------------------------ """
""" 📌 Usando scikit-learn (TfidfVectorizer) """
""" ------------------------------------------------------------------------ """

""" Cuando el flujo de trabajo incluye la creación de vectores de
características para modelos de Machine Learning, puedes aprovechar los
vectorizadores de scikit-learn, que ofrecen una integración práctica con
stopwords.
"""

# Dos documentos de ejemplo
corpus = [
    "En el campo del Procesamiento de Lenguaje Natural, la eliminación de stopwords puede mejorar el rendimiento en diversas tareas.",
    "In the field of Natural Language Processing, removing stopwords can improve performance in many tasks."
]

# stop_words='english' aplica una lista interna de stopwords en inglés
vectorizer = TfidfVectorizer(stop_words='english', lowercase=True)
X = vectorizer.fit_transform(corpus)

print("--- INGLÉS -- EJEMPLO STOPWORDS CON TfidfVectorizer (scikit-learn) ---")
print("Vocabulary (EN):", vectorizer.get_feature_names_out())
print("TF-IDF matrix shape:", X.shape)

"""Comentarios

El TfidfVectorizer convierte el texto en vectores TF-IDF, ignorando stopwords si
se especifica stop_words='english'.
Para usar un filtro en otro idioma o uno personalizado, puedes pasar una lista
de palabras a stop_words."""

stopwords_propias = ["en", "el", "la", "de", "y", "a", "del"]

vectorizer2 = TfidfVectorizer(stop_words=stopwords_propias, lowercase=True)
Y = vectorizer2.fit_transform(corpus)

print("\n\n--- ESPAÑOL -- EJEMPLO STOPWORDS CON TfidfVectorizer (scikit-learn) y lista PROPIA ---")
print("Vocabulario (ES):", vectorizer2.get_feature_names_out())
print("TF-IDF matrix shape:", Y.shape)
```

### Etiquetas POS en spaCy (UPOS)

Las etiquetas POS en spaCy usan el esquema **Universal POS (UPOS)** de Universal Dependencies. Aquí está su significado:

- **ADJ**: Adjetivo - Describe o modifica un sustantivo.
- **ADP**: Adposición - Preposición o posposición.
- **ADV**: Adverbio - Modifica un verbo, adjetivo u otro adverbio.
- **AUX**: Auxiliar - Verbo auxiliar (ej. "haber", "ser").
- **CCONJ**: Conjunción coordinante - Une elementos del mismo nivel (ej. "y", "o").
- **DET**: Determinante - Introduce o especifica un sustantivo (ej. "el", "un").
- **INTJ**: Interjección - Expresión de emoción (ej. "¡ay!", "¡oh!").
- **NOUN**: Sustantivo - Nombre de persona, lugar, cosa o idea.
- **NUM**: Numeral - Número o cantidad (ej. "dos", "100").
- **PART**: Partícula - Palabra invariable con función gramatical (ej. "no").
- **PRON**: Pronombre - Sustituye a un sustantivo (ej. "yo", "ella").
- **PROPN**: Nombre propio - Nombre específico (ej. "Juan", "Madrid").
- **PUNCT**: Puntuación - Signos como coma, punto, etc.
- **SCONJ**: Conjunción subordinante - Introduce una cláusula subordinada (ej. "que", "si").
- **SYM**: Símbolo - Carácteres como %, $, @.
- **VERB**: Verbo - Acción, estado o proceso.
- **X**: Otro - Categoría residual para palabras no clasificadas.

Nota: Estas son etiquetas generales y universales. Para detalles específicos del idioma (XPOS), spaCy depende del modelo usado.

### Etiquetas POS en Stanza (UPOS y XPOS)

Las etiquetas **XPOS** en Stanza corresponden a la etiqueta morfológica o categoría gramatical específica para cada idioma, en contraste con **UPOS**, que representa la etiqueta universal (VERB, NOUN, ADJ, etc.). En el caso de **español**, Stanza está entrenado sobre corpus de Universal Dependencies (por ejemplo, **UD_AnCora** o **UD_GSD**), los cuales incluyen etiquetas XPOS que suelen seguir un esquema detallado de características morfológicas.

Por ejemplo, podrías encontrarte con una XPOS como `NCMS000`, que se desglosa así (según la notación de AnCora/EAGLES):

- **N**: Nombre (Noun).  
- **C**: Común (Common).  
- **M**: Masculino (Masculine).  
- **S**: Singular.  
- **000**: Indica que no hay especificaciones adicionales (por ejemplo, grado, caso, etc.).

Otro ejemplo podría ser un verbo como `VMIP3S0`:

- **V**: Verbo (Verb).  
- **M**: Principal (Main verb).  
- **I**: Indicativo (Indicative).  
- **P**: Presente (Present tense).  
- **3**: Tercera persona (3rd person).  
- **S**: Singular.  
- **0**: Sin rasgo adicional (0).

Estas secuencias provienen de convenciones como **EAGLES** o **AnCora**, donde cada letra o número en la etiqueta expresa un rasgo morfológico específico (tiempo verbal, persona, número, género, modo, etc.).  

En resumen:

1. **UPOS (Universal POS)**: Etiqueta universal y más amplia (p.ej. NOUN, VERB, ADJ).  
2. **XPOS**: Etiqueta detallada y específica del idioma, que codifica múltiples **rasgos morfológicos** (por ejemplo, género, número, tiempo verbal, etc.).  

Para mayor detalle sobre cada uno de los subcomponentes de las etiquetas XPOS en español, puedes consultar la documentación de la **Universal Dependencies** (UD) para español o la descripción de las etiquetas en el corpus **AnCora**.

## 🔹 Sección 4: Análisis Sintáctico de Dependencias

#### 🔍 **Objetivo**  
Comprender cómo las palabras se **relacionan** en una oración mediante **árboles** de dependencias, identificando qué términos actúan como núcleos (verbos) y cuáles son sus componentes, complementos y modificadores.

---

#### 📌 **Introducción**

El **análisis sintáctico de dependencias** muestra cómo las palabras se **relacionan** en una oración. Cada token apunta a su *cabeza* gramatical, formando un **árbol de dependencia**. Esto es esencial para tareas avanzadas de **PLN**, como:
- **Análisis de oraciones complejas** (por ejemplo, detección de sujeto, objeto, etc.).  
- **Extracción de relaciones** (quién hizo qué y a quién).

---

### 🟢 **1. Análisis Sintáctico usando NLTK**

En esta primera aproximación, se muestra cómo **definir una gramática simple (CFG)** y utilizar el **ChartParser** de NLTK para generar **árboles de análisis**. Esto permite ver, paso a paso, la **estructura** de oraciones tanto en **inglés** como en **español**.

**Comentarios (NLTK y CFG)**:
- Define reglas gramaticales: `S -> NP VP`, `NP -> DT NN`, etc.  
- **ChartParser** crea árboles de análisis sintáctico con base en esas reglas.  
- Permite **tokenizar** la oración y luego compararla contra la gramática para generar **posibles derivaciones**.

---

### 🟡 **2. Análisis Sintáctico usando spaCy**

Para un enfoque **estadístico**, spaCy proporciona un pipeline completo que incluye:

1. **Tokenización**.  
2. **Etiquetado gramatical (POS tagging)**.  
3. **Análisis de dependencias**.  

Este último asigna una **relación sintáctica** a cada token (`token.dep_`), e indica cuál es la **palabra cabeza** (`token.head`). Aquí vemos ejemplos en **español** e **inglés**, junto con la interpretación de las etiquetas de dependencia (`nsubj`, `obj`, `amod`, etc.).

**Comentarios (spaCy)**:
- Rápido y robusto, con modelos específicos (`es_core_news_sm`, `en_core_web_sm`).  
- Muestra fácilmente la **relación entre tokens** y su “gobernador” gramatical.  
- Incluye herramientas de **visualización** como `displacy` que permiten ver el árbol de dependencias de forma gráfica.

---

### 🔵 **3. Análisis Sintáctico usando Stanza**

**Stanza** (basado en Stanford NLP) se enfoca en el **soporte multilingüe** mediante modelos que siguen las **Universal Dependencies**. Al procesar oraciones en español e inglés:
- `word.deprel`: tipo de dependencia (p.ej., `nsubj`, `obj`).  
- `word.head`: posición del token padre dentro de la oración; si es `0`, se trata de la **raíz** (`ROOT`).  

**Comentarios (Stanza)**:
- Ideal si requieres **alto soporte multilingüe** y **análisis morfológico** detallado.  
- Ofrece un pipeline completo, similar a spaCy, pero con un enfoque en **investigación** y **multilingüismo**.  
- Las etiquetas siguen el estándar de **Universal Dependencies**, facilitando la comparación entre idiomas.

---

### ✨ **4. Conclusiones**

1. **NLTK (CFG)**:  
   - **Ventaja**: Claridad para ver reglas gramaticales y estructuras formales.  
   - **Desventaja**: Se requiere un conocimiento de gramáticas formales y la definición manual de reglas; no es tan práctico para textos complejos.

2. **spaCy**:  
   - **Ventaja**: Pipeline integrado (POS, NER, dependencias), muy rápido y escalable.  
   - **Desventaja**: Necesita modelos pre-entrenados para cada idioma.

3. **Stanza (Stanford NLP)**:  
   - **Ventaja**: **Multilingüe** y basado en **Universal Dependencies**.  
   - **Desventaja**: Suele ser menos veloz en grandes volúmenes de datos que spaCy.

En general, la elección de la herramienta depende de si buscas un **enfoque formal** (definiendo reglas) o uno **estadístico** (modelos ya entrenados). El análisis de dependencias facilita la comprensión de **quién hace qué a quién** en una oración, y resulta fundamental en aplicaciones de **chatbots**, **extracción de relaciones**, **traducción** y muchos otros casos de uso en **PLN**.  


```python
""" ------------------------------------------------------------------------ """
""" ⚠️ Análisis Sintáctico usando NLTK """
""" ------------------------------------------------------------------------ """

# -----------------------------------
# Ejemplo en inglés
# -----------------------------------

# Definimos una gramática simple en inglés (CFG)
english_grammar = CFG.fromstring("""
    S -> NP VP
    NP -> DT NN | NN
    VP -> VB ADVP
    ADVP -> RB
    DT -> 'The' | 'the'
    NN -> 'cat' | 'runs'
    VB -> 'runs'
    RB -> 'fast'
""")

# Creamos el parser con la gramática
english_parser = ChartParser(english_grammar)

# Oración en inglés
english_sentence = "The cat runs fast"
print("Análisis sintáctico en inglés:")
print(f"Oración: {english_sentence}")

# Tokenizamos la oración
english_tokens = nltk.word_tokenize(english_sentence)

# Generamos y mostramos los árboles de análisis
for tree in english_parser.parse(english_tokens):
    print("\nÁrbol de dependencias:")
    print(tree)
    # Visualizamos el árbol gráficamente
    tree.pretty_print()  # Muestra en consola

# -----------------------------------
# Ejemplo en español
# -----------------------------------

# Definimos una gramática simple en español (CFG)
spanish_grammar = CFG.fromstring("""
    S -> NP VP
    NP -> DT NN | NN
    VP -> VB ADVP
    ADVP -> RB
    DT -> 'El' | 'el'
    NN -> 'gato' | 'corre'
    VB -> 'corre'
    RB -> 'rápido'
""")

# Creamos el parser con la gramática
spanish_parser = ChartParser(spanish_grammar)

# Oración en español
spanish_sentence = "El gato corre rápido"
print("\nAnálisis sintáctico en español:")
print(f"Oración: {spanish_sentence}")

# Tokenizamos la oración
spanish_tokens = nltk.word_tokenize(spanish_sentence)

# Generamos y mostramos los árboles de análisis
for tree in spanish_parser.parse(spanish_tokens):
    print("\nÁrbol de dependencias:")
    print(tree)
    # Visualizamos el árbol gráficamente
    tree.pretty_print()  # Muestra en consola
```
